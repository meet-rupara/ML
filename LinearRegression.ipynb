```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

X = [[1],[3],[5],[7],[9],[4]]
y=[10,30,45,65,80,38]

lr.fit(X,y)

hour =float(input("Enter the number of hours you  studied: "))

mark=lr.predict([[hour]])

print(f"Based on yours {hour}  studied you may score  {mark}")
```
